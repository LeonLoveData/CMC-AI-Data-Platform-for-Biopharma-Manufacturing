# CMC Data Exploration

Exploratory Data Analysis (EDA) for simulated CMC datasets:

- Manufacturing batch data (MES)
- Analytical HPLC results (LIMS)
- Stability study results
- PAT sensor time-series data

This notebook prepares the foundation for feature engineering and ML modeling.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8')

# Paths
BASE_DIR = os.path.dirname(os.getcwd())  # notebooks → project root
RAW_DIR = os.path.join(BASE_DIR, "data", "raw")

batch_file = os.path.join(RAW_DIR, "batch_process.csv")
hplc_file = os.path.join(RAW_DIR, "hplc_results.csv")
stability_file = os.path.join(RAW_DIR, "stability_results.csv")
sensor_file = os.path.join(RAW_DIR, "sensor_timeseries.csv")
```

## Load datasets

```python
batch_df = pd.read_csv(batch_file)
hplc_df = pd.read_csv(hplc_file)
stab_df = pd.read_csv(stability_file)
sensor_df = pd.read_csv(sensor_file)

batch_df.head(), hplc_df.head(), stab_df.head(), sensor_df.head()
```

## Basic dataset info

```python
batch_df.info(), hplc_df.info(), stab_df.info(), sensor_df.info()
```

## Batch Yield Distribution (MES)

```python
plt.figure(figsize=(8,5))
sns.histplot(batch_df["Yield"], kde=True)
plt.title("Batch Yield Distribution")
plt.xlabel("Yield")
plt.show()
```

## Impurity Distribution (LIMS – HPLC)

```python
plt.figure(figsize=(8,5))
sns.histplot(hplc_df["Total_Impurity"], kde=True, color="red")
plt.title("Total Impurity Distribution")
plt.xlabel("Total Impurity (%)")
plt.show()
```

## Stability Study – Potency Over Time

```python
plt.figure(figsize=(8,5))
sns.lineplot(data=stab_df, x="Time_Month", y="Potency", hue="Storage_Temp")
plt.title("Stability Study: Potency vs Time")
plt.show()
```

## Sensor Time-Series (PAT) – Temperature Trend Example

```python
sensor_df["timestamp"] = pd.to_datetime(sensor_df["timestamp"])

plt.figure(figsize=(10,5))
sns.lineplot(data=sensor_df.head(500), x="timestamp", y="Temperature")
plt.title("PAT Sensor Temperature Trend (Sample)")
plt.xticks(rotation=45)
plt.show()
```

## Merge MES + LIMS (Batch-Level Integration)

```python
merged_df = batch_df.merge(hplc_df, on="Batch_ID", how="left")
merged_df.head()
```

## Correlation Heatmap (Batch + Analytical Data)

```python
plt.figure(figsize=(10,6))
sns.heatmap(merged_df.corr(numeric_only=True), annot=False, cmap="coolwarm")
plt.title("Correlation Heatmap – MES + LIMS")
plt.show()
```
